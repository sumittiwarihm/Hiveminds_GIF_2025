```python
from selenium import webdriver
from selenium.webdriver.chrome.service import Service
from selenium.webdriver.chrome.options import Options
from selenium.webdriver.common.by import By
from selenium.webdriver.support.ui import WebDriverWait
from bs4 import BeautifulSoup
import time

# --- Setup Variables ---
scrapped_key = []
scrapped_title = []
scrapped_sponsored = []
scrapped_slot = []
scrapped_page = []
brand = []
scrapped_FSN = []
scrap_price = []
account_name = []
scrapped_mrp=[]
scrapped_brand=[]

# Example acc_kay (replace this with your sliced dict)
# acc_kay = dict(zip(filtered_df['keyword'], filtered_df['account']))
# acc_kay = dict(list(zip(filtered_df['keyword'], filtered_df['account']))[1144:1248])
# Uncomment above line and use your actual `filtered_df`

def get_url_keyword(keyword, page_num):
    url = f"https://www.flipkart.com/search?q={keyword.lower().replace(' ','+')}&otracker=search&otracker1=search&marketplace=FLIPKART&as-show=on&as=off&page={page_num}"
    return url

# Setup ChromeDriver
# chrome_driver_path = r"C:\Users\Rajashekar\Downloads\chromedriver-win64 (5)\chromedriver-win64\chromedriver.exe"  # Change this to your path

options = Options()
options.add_argument("--start-maximized")
# service = Service(executable_path=chrome_driver_path)

# driver = webdriver.Chrome(service=service, options=options)
driver = webdriver.Chrome()
# Scraping starts
iteration = len(acc_kay)

for key, account in acc_kay.items():
    print("Keyword:", key)

    for page in range(1, 2):  # Change range for more pages
        driver.get(get_url_keyword(key, page))
        time.sleep(2)

        html = driver.page_source
        soup = BeautifulSoup(html, "html.parser")

        table = soup.findAll("div", {"class": "_1sdMkc LFEi7Z"})
        if not table:
            table = soup.findAll("div", {"class": "_75nlfW"})

        print("Products found:", len(table))

        for index, value in enumerate(table):
            # Title
            title_tag = value.find("a", {"class": "WKTcLC"}) or value.find("div", {"class": "KzDlHZ"})
            title = title_tag.text.strip() if title_tag else "N/A"
            brand_tag=value.find("div", {"class": "syl9yP"}) 
            brand = brand_tag.text.strip() if brand_tag else "N/A"
            # FSN
            fsn_tag = value.find("a", {"class": "rPDeLR"}) or value.find("a", {"class": "CGtC98"}) or value.find("a", {"class": "tUxRFH"})
            if fsn_tag and fsn_tag.get("href"):
                try:
                    FSN = fsn_tag.get("href").split("?pid=")[1].split("&")[0]
                except:
                    FSN = "N/A"
            else:
                FSN = "N/A"

            # Sponsored Check
            if value.find("div", {"class": "_2ABVdq"}) or value.find("div", {"class": "f8qK5m"}):
                is_sponsored = "Sponsored"
            else:
                is_sponsored = "Organic"

            # Price
            price_tag = value.find("div", {"class": "Nx9bqj"}) or value.find("div", {"class": "Nx9bqj _4b5DiR"})
            price = price_tag.text.replace("₹", "").strip() if price_tag else "0"
            mrp_tag = value.find("div", {"class": "yRaY8j"}) or value.find("div", {"class": "Nx9bqj _4b5DiR"})
            mrp = mrp_tag.text.replace("₹", "").strip() if mrp_tag else "0"
            # Slot
            slot = index + 1

            # Append to lists
            scrapped_key.append(key)
            scrapped_title.append(title)
            scrapped_sponsored.append(is_sponsored)
            scrapped_slot.append(slot)
            scrapped_page.append(page)
            scrapped_FSN.append(FSN)
            scrap_price.append(price)
            account_name.append(account)
            scrapped_mrp.append(mrp)
            scrapped_brand.append(brand)

    print("Remaining:", iteration - 1)
    iteration -= 1

# --- Done ---
driver.quit()
```

```python
flipkart_data=pd.DataFrame(list(zip(scrapped_key,scrapped_title,scrapped_sponsored,scrapped_slot,scrapped_page,scrapped_FSN,scrap_price,account_name,scrapped_mrp,scrapped_brand)),columns=['keyword','title',
                                                                                                  'is_sponsored','slot','page_result','fsn','price','account','mrp','brand'])
```
