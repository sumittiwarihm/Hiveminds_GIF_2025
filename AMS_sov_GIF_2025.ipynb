```python
scrapped_key = []
scrapped_title = []
scrapped_sponsored = []
scrapped_slot = []
scrapped_page = []
scrapped_url = []
scrapped_price=[]
brand_scrapped=[]

def get_url_keyword(keyword,page_num):
    url = f"https://www.amazon.in/s?k={keyword.lower().replace(' ','+')}&page={page_num}&ref=sr_pg_{page_num}"
    return url

url = "https://www.amazon.in"
time.sleep(3)

#try differnt code
try:
    browser.find_element(By.CSS_SELECTOR, "body > div > div.a-row.a-spacing-double-large > div.a-section > div > div > form > div.a-row.a-spacing-large > div > div > div.a-row.a-spacing-base > div > div.a-column.a-span6.a-span-last.a-text-right > a").click()
except:
    try:
        browser.find_element(By.CSS_SELECTOR, "body > div > div.a-row.a-spacing-double-large > div.a-section > div > div > form > div.a-row.a-spacing-large > div > div > div.a-row.a-spacing-base > div > div.a-column.a-span6.a-span-last.a-text-right").click()
    except:
        pass
time.sleep(3)

browser.get(url) 
time.sleep(3)

iteration = len(keyword_list)

for key in keyword_list:

    for page in range(1,2):

        browser.get(get_url_keyword(key,page))
        time.sleep(2)

        soup = BeautifulSoup(browser.page_source, "html.parser")
        
        table = soup.findAll("div",{"class","a-section a-spacing-base"})
        if len(table) == 0:
            table = soup.findAll("div",{"class","a-section a-spacing-base desktop-grid-content-view"})
        if len(table) == 0:
            table = soup.findAll("div",{"class","puis-card-container s-card-container s-overflow-hidden aok-relative puis-expand-height puis-include-content-margin puis puis-v19429uxcrjs2i2lzfaiol3kbc9 s-latency-cf-section puis-card-border"})
        print('len of table',len(table))
        
        for index, value in enumerate(table):
            try:
                brand_span = value.find("span", {"class": "a-size-base-plus a-color-base"})
                brand = brand_span.text.strip() if brand_span else ""
            except:
                brand = 'None'

            try:                                
                title_h2 = value.find("h2", {"class": "a-size-base-plus a-spacing-none a-color-base a-text-normal"})
                title1 = title_h2.text.strip() if title_h2 else ""
            except:
                try:                                
                    title_h2 = value.find("h2", {"class": "a-size-medium a-spacing-none a-color-base a-text-normal"})
                    title1 = title_h2.text.strip() if title_h2 else ""
                except:
                    title1 = 'None'

            # Final combined title
            full_title = f"{brand} {title1}".strip()

            try:
                is_sponsored = value.find("span", {"class": "puis-label-popover-default"}).text
            except:
                try:
                    is_sponsored = value.find("span", {"class": "puis-label-popover-default"}).text
                except:
                    is_sponsored = "Organic"

            try:
                url = value.find('a', {"class": "a-link-normal s-line-clamp-2 s-line-clamp-3-for-col-12 s-link-style a-text-normal"}).get('href')
                url = "https://www.amazon.in" + url
            except:                                 
                try:                                  
                    url = value.find('a', {"class": "a-link-normal s-line-clamp-3 s-link-style a-text-normal"}).get('href')
                    url = "https://www.amazon.in" + url
                except:
                    try:                                  
                        url = value.find('a', {"class": "a-link-normal s-line-clamp-4 s-link-style a-text-normal"}).get('href')
                        url = "https://www.amazon.in" + url
                    except:
                        url = ""
                

            try:
                price = value.find("span", {"class": "a-price-whole"}).text
            except:
                try:
                    price = value.find("span", {"class": "a-offscreen"}).text
                except:
                    price = "0"

            slot = index + 1

            # Append to lists
            scrapped_key.append(key)
            scrapped_title.append(full_title)
            scrapped_sponsored.append(is_sponsored)
            scrapped_slot.append(slot)
            scrapped_page.append(page)
            scrapped_url.append(url)
            scrapped_price.append(price)
            brand_scrapped.append(brand)
        
    print("Remaining",iteration)
    iteration = iteration - 1
```

```python
data = pd.DataFrame(list(zip(scrapped_key,scrapped_title,scrapped_sponsored,scrapped_slot,scrapped_page,scrapped_url,scrapped_price,brand_scrapped)),
                    columns=['keyword','title','is_sponsored','slot','Page Result','url','price','brand'])

data['date'] = date.today()

data_g = data[['keyword', 'title', 'is_sponsored', 'slot', 'Page Result','date','url','price','brand']]
# data_g.sample(3)
```
