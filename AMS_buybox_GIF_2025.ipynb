```python
data_1['Free_Delivery'] = pd.to_datetime(data_1['Free_Delivery'], errors='coerce', format='%d %b', exact=False) \
                  .combine_first(pd.to_datetime(data_1['Free_Delivery'], errors='coerce', format='%B %d', exact=False))

data_1['Fast_Delivery'] = pd.to_datetime(data_1['Fast_Delivery'], errors='coerce', format='%d %b', exact=False) \
                  .combine_first(pd.to_datetime(data_1['Fast_Delivery'], errors='coerce', format='%B %d', exact=False))


current_year = datetime.datetime.now().year
data_1['Free_Delivery'] = data_1['Free_Delivery'].apply(lambda x: x.replace(year=current_year))
data_1['Fast_Delivery'] = data_1['Fast_Delivery'].apply(lambda x: x.replace(year=current_year))

data_1['Free_Delivery'] = data_1['Free_Delivery'].dt.date
data_1['Fast_Delivery'] = data_1['Fast_Delivery'].dt.date

data_1['Date'] = datetime.date.today()
data_1.head(2)
```
